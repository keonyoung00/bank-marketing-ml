
# 04. Model Experiments

## 1. Introduction

### 1.1 Objective & Scope

The objective of this notebook is to improve upon the Logistic Regression baseline established in `03-baseline-modeling.ipynb` and select a final model configuration through systematic experimentation.

The business objective remains to rank customers by their predicted probability of subscribing to a term deposit, allowing limited call-center resources to prioritize the most promising customers.

This notebook focuses on:

- Evaluating alternative feature representations and preprocessing strategies.
- Comparing different model families.
- Optimizing promising models through hyperparameter tuning.
- Analyzing model behavior, interpretability, and robustness.
- Selecting the final model configuration based on validation performance.

All experiments use the established chronological training and validation sets.

The test set remains frozen for final evaluation in `05-final-evaluation.ipynb`.

### 1.2 Baseline Recap

The previous notebook established a Logistic Regression baseline using the following configuration:

- **Model:** `LogisticRegression(max_iter=1000)`
- **Numeric preprocessing:** `StandardScaler`, including raw `pdays`
- **Categorical preprocessing:** `OneHotEncoder(handle_unknown="ignore")`
- **Features:** 19 predictors, excluding `duration`
- **Data split:** Chronological 70% / 20% / 10%

The baseline achieved the following validation performance:

| Metric | Baseline |
|---|---:|
| Average Precision | 0.297231 |
| ROC-AUC | 0.692923 |
| Precision@10% | 0.3859 |
| Recall@10% | 0.2785 |
| Lift@10% | 2.7839 |

These results serve as the reference for subsequent experiments.

Average Precision (AP) is the primary model-selection metric because the main objective is to identify and prioritize potential subscribers.

ROC-AUC and business-oriented ranking metrics provide complementary evidence about model performance.

### 1.3 Experiment Plan

The experiments will be organized into the following stages.

**Stage 1 — Feature & Preprocessing Experiments**

- Evaluate selected alternative feature representations.
- Investigate potentially useful feature engineering and selection strategies.
- Compare preprocessing configurations when justified.

**Stage 2 — Model Comparison**

- Compare Logistic Regression with alternative model families.
- Initially consider Random Forest and Gradient Boosting models.
- Use model-appropriate preprocessing while maintaining consistent data splits and evaluation metrics.

**Stage 3 — Hyperparameter Optimization**

- Identify promising candidate models.
- Define reasonable hyperparameter search spaces.
- Evaluate tuned configurations using the validation set.

**Stage 4 — Model Analysis**

- Analyze model errors and important predictors.
- Investigate sensitivity to temporal distribution shift.
- Compare business-oriented ranking performance at different contact capacities.

**Stage 5 — Final Model Selection**

- Summarize experimental results.
- Select the final model and preprocessing configuration.
- Document the selection rationale and remaining limitations.

### Evaluation Principles

The following principles apply throughout this notebook:

1. All learned preprocessing is fitted exclusively on the training set during development.
2. All candidate models are evaluated on the same chronological validation set.
3. Average Precision is the primary model-selection metric.
4. ROC-AUC, Precision@K, Recall@K, and Lift@K are used as supporting metrics.
5. The test set is not used for experimentation, tuning, or model selection.
6. Additional model complexity is introduced only when supported by validation evidence.

The selected configuration will be carried forward to `05-final-evaluation.ipynb` for final refitting and frozen test evaluation.
